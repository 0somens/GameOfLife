```python
import pandas as pd
```

```python
# /workspaces/GameOfLife/data Ruta
try:
    df = pd.read_csv('/workspaces/GameOfLife/data/KtoCsv.csv', sep=';')
except UnicodeDecodeError:
    print("Error: Unable to decode the file. Please check the file encoding.")
df.head()
```

```python
x = df['Celcius'].values
y = df['Kelvin'].values
```

```python
import tensorflow as tf
# from tensorflow import keras
```

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.models import Sequential # ?
from tensorflow.keras.layers import Dense     # ?
# import numpy as np

modelo = Sequential([
    Dense(units= 1, input_shape=[1]) #Una sola neurona
])
# modelo.compile(optimizer='adam', loss='mean_squeared_error')
modelo.compile(
    optimizer=tf.keras.optimizers.Adam(0.1), # Va a hacer los ajustados, podría ser que los datos tuvieran varianza 
    loss='mean_squared_error'
)
# modelo.compile(optimizer='sgd', loss='mean_squared_error') # sgd Stochastic Gradient Descent
```

```python
modelo.fit(x, y, epochs=500, verbose=1) # verbose=0 no muestra nada, verbose=1 muestra la barra de progreso
```

```python
temperatura = 60
prediccion = modelo.predict([temperatura])
print(prediccion)
```
